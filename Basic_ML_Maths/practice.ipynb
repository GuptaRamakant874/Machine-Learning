```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

from sklearn.datasets import fetch_california_housing

df = fetch_california_housing()
```

```python
df
```

```python
dataset = pd.DataFrame(df.data)
```

```python
dataset.columns = df.feature_names
dataset.head(10)
```

```python
## train test split

from sklearn.model_selection import train_test_split

X = dataset
y = df.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)

```

```python
X_train
```

# Regression Implementation

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_train
```

```python
X_test = scaler.transform(X_test)
X_test
```

```python
## import Linear Regression

from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train,y_train)

## cross validation -- mean that model will train on different set of data
from sklearn.model_selection import cross_val_score 
# regression,x_train,y_train,scoring='which type of value you want to calculate, cv=how much different data will train
mse = cross_val_score(regression,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
np.mean(mse)
```

```python
# prediction
reg_predict = regression.predict(X_test)
```

```python
reg_predict
```

```python
#comparing with truth value
import seaborn as sns
sns.displot(reg_predict-y_test,kind='kde')
```
