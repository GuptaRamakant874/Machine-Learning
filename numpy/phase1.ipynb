### Creating Array from numpy

```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,6])
print(arr1)
```

```python
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr2)
```

### List vs Numpy array

```python

pylist = [1,2,3]
print("python list multiplication",pylist*2)

np_arr = np.array([1,2,3]) #Element wise multiplication
print("py numpy multiplication",np_arr*2)

```

```python

import time
start = time.time()
list = [i*2 for i in range(1000000)]
print(" List operation time:",time.time()-start)


start = time.time()
nparr = np.arange(1000000)*2
print(" Numpy operation time:",time.time()-start)
```

### creating array from scratch

```python
zeros = np.zeros((3,4)) #(row,col)
print(zeros)

ones = np.ones((2,3))
print(ones)

full = np.full((2,2),7) #((row,col),value)
print("full Array: \n",full)

ran = np.random.random((2,3))
print("random array: \n",ran)

sequence = np.arange(0,9,1) #(start, end, gap between values)
print("sequence array: \n",sequence)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1,2,3])
print("vector : ",vector)

matrix = np.array([[1,2,3],[4,5,6]])
print("matrix : ",matrix)

tensor = np.array([[[1,2] , [3,4]],
                   [[5,6] , [7,8]]])
print("tensor : ",tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape ",arr.shape)
print("Dimension ",arr.ndim)
print("Size ", arr.size)
print("Dtype ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("original Array: ", arr)

reshape = arr.reshape((3,4))
print("\n Reshaped array ",reshape)

flatten = reshape.flatten() #this return original array
print("\n flattened array ",flatten)

raveled = reshape.ravel() # this return original array
print("\n raveled array ", raveled)

transpose = reshape.T
print("\n transposed array ",transpose)
```
