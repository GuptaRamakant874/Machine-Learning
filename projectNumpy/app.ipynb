# Data Manipulation


```python
import pandas as pd

df = pd.read_csv('titanic.csv')

print(df.shape)

print(df.columns)
```

```python
import matplotlib.pyplot as plt

team_counts = df['venue'].value_counts()

plt.figure(figsize=(10,5))
plt.bar(team_counts.index, team_counts.values)
plt.title("Matches Played by Each Team")
plt.xlabel("venue")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['batting_team'], bins=15)
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import seaborn as sns

display(df.head())
sns.heatmap(df.isnull(), yticklabels=False, cmap='viridis')
sns.countplot(x='Survived',data=df)
plt.show()
df['Survived'].value_counts()
df['Sex'].value_counts()
ax = sns.countplot(x='Sex', data=df)
plt.title('Passenger Count by Sex')

for p in ax.patches:
    ax.annotate(f'{p.get_height()}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9),
                textcoords='offset points')
plt.show()



```

```python
ax = sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival Count by Sex (0=Not Survived, 1=Survived)')

for p in ax.patches:
    ax.annotate(f'{p.get_height()}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9),
                textcoords='offset points')
plt.show()
```

```python
ax = sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival Count by Passenger Class (0=Not Survived, 1=Survived)')

for p in ax.patches:
    ax.annotate(f'{p.get_height()}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 9),
                textcoords='offset points')
plt.show()
```

```python
sns.countplot(x='SibSp',data=df)
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), bins='auto', kde=True)
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Fare'], bins='auto', kde=True)
plt.title('Distribution of Passenger Fares')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
